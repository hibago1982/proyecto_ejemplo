```python
# ===========================================================
# 📂 PASO 2: Clonar tu repositorio (debe existir en GitHub)
# Asegúrate de haber creado "proyecto_ejemplo" en github.com/hibago1982
# ===========================================================
!git clone https://github.com/hibago1982/proyecto_ejemplo.git
%cd proyecto_ejemplo
print("✅ Repositorio clonado y carpeta activada.")
```

```python
# ===========================================================
# 📝 PASO 3: Crear un archivo de ejemplo (reporte simple)
# ===========================================================
contenido_reporte = """\
# Reporte Diario de Producción
Fecha: 2025-11-12
Línea: L1
Unidades producidas: 1,250
Paradas programadas: 0
Paradas no programadas: 15 min
Estado: ✅ Operación estable
"""

# Guardamos el archivo
with open("reporte_dia1.txt", "w", encoding="utf-8") as f:
    f.write(contenido_reporte)

print("✅ Archivo 'reporte_dia1.txt' creado.")
!ls  # Muestra los archivos para confirmar
```

```python
# ===========================================================
# 🔐 PASO 4: Configurar autenticación con GitHub
# Esto evita el error: "fatal: could not read Username..."
#
# ⚠️ NUNCA escribas el token directamente en el notebook.
# Un notebook se sube a GitHub con todo su contenido: un token
# escrito aquí queda publicado y hay bots que rastrean GitHub
# buscando exactamente eso. Aquí se pide en tiempo de ejecución,
# así que no queda guardado dentro del archivo .ipynb.
# ===========================================================
import subprocess
from getpass import getpass

GITHUB_USER = "hibago1982"
REPO = "proyecto_ejemplo"

# En Colab puedes guardar el token una sola vez en el panel de
# Secretos (🔑) con el nombre GITHUB_TOKEN y se leerá solo.
# Si no está disponible, se pide por teclado sin mostrarlo en pantalla.
try:
    from google.colab import userdata
    GITHUB_TOKEN = userdata.get("GITHUB_TOKEN")
except Exception:
    GITHUB_TOKEN = getpass("Token de GitHub (no se mostrará al escribir): ")

url = f"https://{GITHUB_USER}:{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{REPO}.git"

# Se ejecuta con subprocess y no con !git: así el token viaja como
# argumento directo, sin pasar por el shell ni por el historial de la celda.
subprocess.run(["git", "remote", "set-url", "origin", url], check=True)

print("✅ Autenticación con GitHub configurada.")
```

```python
# ===========================================================
# 📦 PASO 5: Añadir, commitear y subir los cambios
# ===========================================================
!git add reporte_dia1.txt
!git commit -m "Agregué primer reporte diario de producción"

# ¡Ahora sí! El push funcionará sin pedir usuario/contraseña
!git push origin main

print("\n🎉 ¡Éxito! Tu archivo ya está en GitHub.")
print("👉 Ve a: https://github.com/hibago1982/proyecto_ejemplo para verlo.")
```
